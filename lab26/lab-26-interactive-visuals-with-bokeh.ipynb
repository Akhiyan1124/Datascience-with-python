```python
import pandas as pd
import numpy as np

from bokeh.plotting import figure, show, output_file
from bokeh.io import output_notebook
from bokeh.models import HoverTool
from bokeh.layouts import column, gridplot
from bokeh.palettes import Category10

from datetime import datetime, timedelta
```

```python
output_notebook()
```

```python
x = np.linspace(0, 4*np.pi, 100)
y = np.sin(x)

p = figure(
    title="Basic Interactive Plot",
    width=800,
    height=400,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

p.line(x, y, line_width=2, color='blue', legend_label='sin(x)')

p.circle(
    x[::10],
    y[::10],
    size=8,
    color='red',
    alpha=0.6
)

p.legend.click_policy = "hide"

show(p)
```

```python
np.random.seed(42)

data = pd.DataFrame({
    'x': np.random.randn(100),
    'y': np.random.randn(100),
    'value': np.random.randint(5, 30, 100)
})

p = figure(
    title="Scatter Plot with Hover",
    width=700,
    height=400
)

hover = HoverTool(tooltips=[
    ("X Value", "@x"),
    ("Y Value", "@y"),
    ("Size", "@value")
])

p.add_tools(hover)

p.circle(
    'x',
    'y',
    size='value',
    alpha=0.6,
    source=data
)

show(p)
```

```python
dates = pd.date_range(
    start='2024-01-01',
    periods=365
)

prices = np.cumsum(
    np.random.normal(0, 2, 365)
) + 150

volumes = np.random.randint(
    1000000,
    5000000,
    365
)

stock_data = pd.DataFrame({
    'date': dates,
    'price': prices,
    'volume': volumes
})

stock_data.head()
```

```python
p = figure(
    title="Interactive Stock Price Chart",
    x_axis_type='datetime',
    width=900,
    height=400,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

hover = HoverTool(tooltips=[
    ("Date", "@date{%F}"),
    ("Price", "@price{0.00}"),
    ("Volume", "@volume{0,0}")
],
formatters={
    '@date': 'datetime'
})

p.add_tools(hover)

p.line(
    'date',
    'price',
    source=stock_data,
    line_width=2,
    color='blue'
)

p.circle(
    'date',
    'price',
    source=stock_data,
    size=5,
    color='red'
)

show(p)
```

```python
weather_data = pd.DataFrame({
    'date': dates,
    'temperature': np.random.normal(25, 5, 365),
    'humidity': np.random.normal(60, 10, 365),
    'wind_speed': np.random.normal(15, 4, 365)
})

weather_data.head()
```

```python
temp_chart = figure(
    title="Temperature Over Time",
    x_axis_type='datetime',
    width=700,
    height=300
)

temp_chart.line(
    weather_data['date'],
    weather_data['temperature'],
    line_width=2,
    color='orange'
)

humidity_chart = figure(
    title="Humidity Levels",
    x_axis_type='datetime',
    width=700,
    height=300
)

humidity_chart.vbar(
    x=weather_data['date'],
    top=weather_data['humidity'],
    width=0.5,
    alpha=0.6
)

layout = column(
    temp_chart,
    humidity_chart
)

show(layout)
```

```python
from bokeh.models import Slider
from bokeh.layouts import column

slider = Slider(
    start=0,
    end=100,
    value=50,
    step=1,
    title="Example Slider"
)

layout = column(slider, p)

show(layout)
```

```python
from bokeh.models import LinearColorMapper
from bokeh.transform import linear_cmap
from bokeh.palettes import Viridis256

mapper = linear_cmap(
    field_name='price',
    palette=Viridis256,
    low=stock_data['price'].min(),
    high=stock_data['price'].max()
)

p = figure(
    title="Color Mapping Example",
    x_axis_type='datetime',
    width=800,
    height=400
)

p.circle(
    'date',
    'price',
    source=stock_data,
    size=10,
    color=mapper,
    alpha=0.7
)

show(p)
```
