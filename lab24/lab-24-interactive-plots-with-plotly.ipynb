## 📊 Task 3: Interactive Bar Charts

```python
products = ['Laptops', 'Smartphones', 'Tablets', 'Headphones', 'Cameras']

q1_sales = [45000, 62000, 28000, 15000, 22000]
q2_sales = [52000, 68000, 32000, 18000, 25000]
q3_sales = [48000, 71000, 30000, 20000, 28000]

fig_bar = go.Figure()

fig_bar.add_trace(go.Bar(name='Q1', x=products, y=q1_sales))
fig_bar.add_trace(go.Bar(name='Q2', x=products, y=q2_sales))
fig_bar.add_trace(go.Bar(name='Q3', x=products, y=q3_sales))

fig_bar.update_layout(barmode='group')

fig_bar.show()
```

## 📊 Task 4: Dynamic Axes

```python
import datetime

start_date = datetime.date(2023, 1, 1)
dates = [start_date + datetime.timedelta(days=x) for x in range(365)]

np.random.seed(42)
prices = np.cumsum(np.random.normal(0, 2, 365)) + 100

fig_dynamic = go.Figure()

fig_dynamic.add_trace(go.Scatter(
    x=dates,
    y=prices,
    mode='lines',
    name='Stock Price'
))

fig_dynamic.update_layout(
    xaxis=dict(
        rangeselector=dict(
            buttons=[
                dict(count=7, label="7d", step="day", stepmode="backward"),
                dict(count=30, label="30d", step="day", stepmode="backward"),
                dict(step="all")
            ]
        ),
        rangeslider=dict(visible=True),
        type="date"
    )
)

fig_dynamic.show()
```
